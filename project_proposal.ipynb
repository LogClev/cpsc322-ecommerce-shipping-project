### **Dataset Description**

For this project, we'll be using the **_E-Commerce Shipping Data_** dataset published by Prachi Gopalani on _kaggle.com_. The current version of this dataset was published in 2020.   
Link to dataset: [Click here!](https://www.kaggle.com/datasets/prachi13/customer-analytics?)

**Description:** Data is organized using CSV format. This dataset includes 12 attributes and 10999 instances and contains information regarding shipped products for an e-commerce company that sells electronic products. Each instance is a separate product order.  

**Attributes:**
1. *ID* (Int) -> the unique ID of the customer
2. *Warehouse_block* (String) -> the alphabetical block region of the e-commerce company warehouse that product originated in
3. *Mode_of_Shipment* (String) -> how the product was shipped (by ship, flight, and road)
4. *Customer_care_calls* (Int) -> number of calls customer made to inquire about the shipment status
5. *Customer_rating* (Int) -> ratings the company recieved from the customer (1 being lowest and 5 being highest)
6. *Cost_of_the_Product* (Int) -> cost of the product shipped in U.S. dollars
7. *Prior_purchases* (Int) -> the number of times the customer has purchased from the e-commerce company before
8. *Product_importance* (String) -> a company made parameter to evaluate the importance of a product being recieved by the customer
9. *Gender* (String) -> binary categorization of customer gender (male or female)
10. *Discount_offered* (Int) -> discount amount in U.S. dollars offered on the product
11. *Weight_in_gms* (Int) -> the total weight of the product being shipped in grams
12. *Reached.on.Time_Y.N* (Int) -> binary classification of whether the product arrived to the customer on time (1 indicates late arrival while 0 indicates an on-time arrival)

**What will we try to predict for this project?:** Our attribute of interest that we will be trying to predict is *Reached.on.Time_Y.N*. This is a binary classification between the classes *On Time* (0) and *Not On Time* (1). We will utilize the remaining attributes to search for trends that will aid us in predicting whether or not a product shipped by this company will arrive to the customer on time.


### **Implementation and Merit**

For this project, there are a few **challenges** we anticipate to face:
1. Preventing data leakage: The datatset contains attributes like *Customer_care_calls* and *Customer_rating* which are likely to be noted only AFTER product delivery. For our project, we want to analyze models the e-commerce company can use to predict product delivery (on time/not on time) status BEFORE a product is shipped.

2. Handling class distribution: The dataset contains significantly more instances belonging to the *Not On Time* class compared to the *On Time* class. We'll need to evaluate ways we can minimize model reliance on favoring the majority class when making predictions.

3. Encoding features: A few attributes are currently categorical: *Warehouse_block*, *Mode_of_Shipment*, *Product_importance* and *Gender*. If we decide to use models that rely on learning from numerical attributes (like kNN) we will need to appropriately encode these attributes before we can use them in our models.

Regarding **feature selection**, our dataset only contains 12 attributes which is a reasonable number to work with without extensive pruning. However, we will still evaluate feature importance and make selections:
- The *ID* attribute is trivial for the predictions we are trying to make as it is just an instance identifier. We will likely remove this attribute.

- Depending on how we choose to approach the above stated data leakage issue, we may also remove the *Customer_care_calls* and *Customer_rating* attributes to preserve the integrity of our models.

- Additionally, we will conduct intial EDA to search for relationships between features and the predicted classes (distributions, summary statistics, charts, etc.). This will help inform any further feature selection decisions we make.
